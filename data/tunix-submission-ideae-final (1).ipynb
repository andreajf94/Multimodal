## 1. Introduction

Chain-of-thought prompting has shown that encouraging language models to "show their work" improves reasoning performance. However, prompted reasoning lacks consistency; models may ramble, skip logical steps, or produce correct reasoning leading to incorrect answers. This work introduces a training methodology for instilling *structured reasoning* in language models using the IDEA-E framework. Using reinforcement learning with Group Relative Policy Optimization (GRPO) on Google Tunix, we fine-tune Gemma-2-2B-IT to follow this structure consistently. 

Our approach is built around four key components:

1. **IDEA-E framework design**: Adapted from Trillium Health Partners’ IDEA Ethical Decision-Making Framework, a structured process originally developed in healthcare to navigate complex ethical dilemmas. We reframe it for language models as a five-stage scaffold (Identify, Define, Ethics, Act, Explain) that enforces reflection before action, with separated reasoning and answer blocks for independent evaluation.

2. **Large-scale dataset construction**: We generate and curate 27,198 teacher-distilled examples from nine reasoning benchmarks, using DeepSeek-Reasoner as the primary teacher model with GPT-4o-mini for review. The dataset includes strict correctness filtering, difficulty tagging, and license management.


3. **Reward function engineering**: We iteratively develop and test reward components, culminating in a multi-component function that combines answer correctness, format compliance, section presence, and TF-IDF-based reasoning similarity for dense supervision.

4. **Curriculum sampling**: We implement a four-phase sampling strategy to stabilize training across a mixed-benchmark dataset, incorporating difficulty ramping, weak-bucket remediation, and replay buffering.

## 2. The IDEA-E Framework

### 2.1 Design Motivation

Our goal was to make reasoning both *repeatable* and *structured*, while ensuring that safety and ethical considerations are not optional. The original IDEA framework was chosen as a foundation because it is a well-established decision-making process that makes ethical reflection an explicit and early part of the reasoning workflow. We add the *Explain* step at the end to prompt the model to summarize the preceding steps before outputting a final answer. Instructions to use IDEA-E tags are included in the system prompt for both the teacher and student models. This encourages the teacher model to follow the desired structure and makes it easier to evaluate outputs, through a combination of heuristics and manual review, before accepting them as training samples.

### 2.2 Framework Structure

The scaffold is designed to walk the model through a minimal, interpretable reasoning path.

```
<reasoning>
[I - Identify] What kind of problem is this? Goal + constraints?
[D - Define]   Lock assumptions, definitions, and success criteria.
[E - Ethics]   Quick scan: OK / Needs constraints / Refuse.
[A - Act]      Pick approach and execute minimum steps.
[E - Explain]  Present answer clearly with justification.
</reasoning>
<answer>
[Final answer]
</answer>
```

Each stage serves a specific purpose: **Identify** forces explicit problem categorization; **Define** locks assumptions early; **Ethics** provides a safety gate before action; **Act** encourages focused solution steps; **Explain** separates justification from the final answer. The separation of `<reasoning>` and `<answer>` blocks enables independent evaluation of reasoning quality and answer correctness.

## 5 Curriculum Sampling

### 5.1 Motivation

Training on the mixed-benchmark dataset using naïve uniform sampling led to high reward variance and early performance plateaus. Because benchmarks differ in answer formats and difficulty levels, the model struggled to converge on a stable output structure. This instability was especially pronounced when difficult samples appeared early in training.

To address these issues, we introduced curriculum-based sampling, which prioritizes output stabilization in early phases and gradually increases reasoning difficulty. This structured progression improves format reliability and reduces variance across benchmarks.

### 5.2 Four-Phase Implementation

| Phase | Steps | Strategy |
|-------|-------|----------|
| A: Format Stabilization | 0–20% | Easy-heavy (70/25/5 easy/med/hard) |
| B: Difficulty Ramp | 20–80% | Linear interpolation to dataset distribution (37/42/21) |
| C: Weak-Bucket Focus | 80–90% | Upweight underperforming benchmarks inversely to reward |
| D: Consolidation | 90–100% | Balanced sampling (37/42/21) + 15% replay buffer |

*Table 2: Curriculum sampling phases*

**Phase A** focuses on reliable IDEA-E output structure before difficulty increases reward sparsity. **Phase B** implements linear difficulty ramp. **Phase C** adjusts benchmark weights inversely proportional to recent reward, providing targeted remediation. **Phase D** uses balanced sampling with replay buffer to reduce catastrophic forgetting.

---

````python
# =============================================================================
# CURRICULUM SAMPLER
# =============================================================================
class CurriculumSampler:
    """Phase-based curriculum sampler for GRPO training."""
    
    def __init__(self, dataset_path, total_steps, seed=42):
        self.total_steps = total_steps
        self.rng = random.Random(seed)
        self.np_rng = np.random.default_rng(seed)
        self.bucket_rewards = defaultdict(list)
        self.replay_buffer = []
        self.max_replay_size = 500
        self.data = self._load_dataset(dataset_path)
        self._organize_by_metadata()
        # Without-replacement sampling tracking
        self.used_indices = {1: set(), 2: set(), 3: set()}  # Track used indices per difficulty
        self.reset_threshold = 0.9  # Reset when 90% of bucket used
    
    def _load_dataset(self, path):
        data = []
        with open(path, newline="", encoding="utf-8") as f:
            reader = csv.DictReader(f)
            for row in reader:
                diff = row.get("difficulty", "2")
                try:
                    row["difficulty_int"] = int(diff) if diff else 2
                except:
                    row["difficulty_int"] = 2
                data.append(row)
        print(f"CurriculumSampler: Loaded {len(data)} samples")
        return data
    
    def _organize_by_metadata(self):
        self.by_benchmark = defaultdict(list)
        self.by_difficulty = defaultdict(list)
        for item in self.data:
            benchmark = item.get("source_benchmark", "unknown")
            difficulty = item.get("difficulty_int", 2)
            self.by_benchmark[benchmark].append(item)
            self.by_difficulty[difficulty].append(item)
        print(f"  Benchmarks: {list(self.by_benchmark.keys())}")
        print(f"  Difficulties: {dict((k, len(v)) for k, v in self.by_difficulty.items())}")
    
    def get_phase(self, step):
        progress = step / self.total_steps
        if progress < PHASE_A_END:
            return "A"
        elif progress < PHASE_B_END:
            return "B"
        elif progress < PHASE_C_END:
            return "C"
        return "D"
    
    def _get_difficulty_weights(self, step):
        progress = step / self.total_steps
        if progress < PHASE_A_END:
            # Phase A: Start easy-heavy for format learning
            return {1: 0.70, 2: 0.25, 3: 0.05}
        elif progress < PHASE_B_END:
            # Phase B: Linear ramp to match dataset distribution
            phase_progress = (progress - PHASE_A_END) / (PHASE_B_END - PHASE_A_END)
            start = {1: 0.70, 2: 0.25, 3: 0.05}
            end = {1: 0.37, 2: 0.42, 3: 0.21}  # Match dataset: 37% easy, 42% medium, 21% hard
            return {k: start[k] + (end[k] - start[k]) * phase_progress for k in [1, 2, 3]}
        # Phases C & D: Use dataset's natural distribution
        return {1: 0.37, 2: 0.42, 3: 0.21}
    
    def _get_benchmark_weights(self, step):
        benchmarks = list(self.by_benchmark.keys())
        n = len(benchmarks)
        weights = {b: 1.0 / n for b in benchmarks}
        
        if self.get_phase(step) == "C" and self.bucket_rewards:
            avg_rewards = {}
            for b in benchmarks:
                rewards = self.bucket_rewards.get(b, [])
                avg_rewards[b] = np.mean(rewards) if rewards else 0.5
            min_r, max_r = min(avg_rewards.values()), max(avg_rewards.values())
            if max_r > min_r:
                for b in benchmarks:
                    norm = (avg_rewards[b] - min_r) / (max_r - min_r)
                    weights[b] = 1.0 - norm + 0.3
        
        for b in benchmarks:
            weights[b] = max(weights[b], MIN_WEIGHT_FLOOR)
        total = sum(weights.values())
        return {b: w / total for b, w in weights.items()}
    
    def sample_batch(self, step, batch_size=1):
        phase = self.get_phase(step)
        batch = []
        for _ in range(batch_size):
            use_replay = (
                self.replay_buffer
                and self.rng.random() < REPLAY_FRACTION
                and phase in ("B", "C", "D")
            )
            if use_replay:
                item = self.rng.choice(self.replay_buffer)
            else:
                item = self._sample_one(step)
                if len(self.replay_buffer) < self.max_replay_size:
                    self.replay_buffer.append(item)
                elif self.rng.random() < 0.1:
                    idx = self.rng.randint(0, len(self.replay_buffer) - 1)
                    self.replay_buffer[idx] = item
            batch.append(item)
        return batch
    
    def _sample_one(self, step):
        diff_weights = self._get_difficulty_weights(step)
        bench_weights = self._get_benchmark_weights(step)
        difficulties = list(diff_weights.keys())
        diff_probs = [diff_weights[d] for d in difficulties]
        chosen_diff = self.np_rng.choice(difficulties, p=diff_probs)
        candidates = self.by_difficulty.get(chosen_diff, self.data)
        if not candidates:
            candidates = self.data
        
        # Filter out already-used samples (without replacement)
        available = []
        available_indices = []
        for i, item in enumerate(candidates):
            if i not in self.used_indices[chosen_diff]:
                available.append(item)
                available_indices.append(i)
        
        # Reset bucket if exhausted or nearly exhausted
        if not available or len(available) < len(candidates) * (1 - self.reset_threshold):
            self.used_indices[chosen_diff].clear()
            available = candidates
            available_indices = list(range(len(candidates)))
            print(f"  Reset difficulty {chosen_diff} bucket at step {step}")
        
        # Sample from available items with benchmark weighting
        if len(available) > 1:
            item_weights = []
            for item in available:
                b = item.get("source_benchmark", "unknown")
                item_weights.append(bench_weights.get(b, 0.1))
            item_weights = np.array(item_weights)
            item_weights /= item_weights.sum()
            idx = self.np_rng.choice(len(available), p=item_weights)
            chosen_item = available[idx]
            chosen_candidate_idx = available_indices[idx]
        else:
            chosen_item = available[0] if available else self.rng.choice(self.data)
            chosen_candidate_idx = available_indices[0] if available_indices else 0
        
        # Mark as used
        self.used_indices[chosen_diff].add(chosen_candidate_idx)
        return chosen_item
    
    def get_stats(self, step):
        return {
            "phase": self.get_phase(step),
            "step": step,
            "difficulty_weights": self._get_difficulty_weights(step),
            "benchmark_weights": self._get_benchmark_weights(step),
            "replay_buffer_size": len(self.replay_buffer),
            "used_counts": {d: len(used) for d, used in self.used_indices.items()},
            "bucket_sizes": {d: len(items) for d, items in self.by_difficulty.items()},
        }


class CurriculumDataset:
    """Wraps CurriculumSampler to provide iterator interface for GRPO trainer."""
    
    def __init__(self, sampler, template, system_prompt, batch_size, total_batches):
        self.sampler = sampler
        self.template = template
        self.system_prompt = system_prompt
        self.batch_size = batch_size
        self.total_batches = total_batches
        self._current_step = 0
    
    def __len__(self):
        return self.total_batches
    
    def __iter__(self):
        self._current_step = 0
        return self
    
    def __next__(self):
        if self._current_step >= self.total_batches:
            raise StopIteration
        
        items = self.sampler.sample_batch(self._current_step, batch_size=self.batch_size)
        
        batch = {
            "prompts": np.array([
                PROMPT_TEMPLATE.format(question=str(item.get("question", "")))
                for item in items
            ], dtype=object),
            "question": np.array([str(item.get("question", "")) for item in items], dtype=object),
            "reasoning": np.array([str(item.get("reasoning", "")) for item in items], dtype=object),
            "answer": np.array([str(item.get("answer") or item.get("gt_final", "")) for item in items], dtype=object),
        }
        
        if self._current_step % 100 == 0:
            stats = self.sampler.get_stats(self._current_step)
            print(f"Step {self._current_step}: Phase {stats['phase']}, Diff weights: {stats['difficulty_weights']}")
        
        self._current_step += 1
        return batch
````

## 6 Reward Function Design

The reward function comprises four components, combined additively as:
\begin{equation}
R_{\text{total}} = R_{\text{answer}} + R_{\text{format}} + R_{\text{IDEA-E}} + R_{\text{reasoning}}.
\end{equation}

The first component, $R_{\text{answer}}$, is correctness computed exclusively from the \texttt{<answer>} block using the tiered scoring system shown below. A robust answer extraction function is used to accommodate different benchmark answer formats:
\begin{align}
R_{\text{answer}} &= 4.0 && \text{for exact matches (Tier 1)} \nonumber\\
R_{\text{answer}} &= 3.5 && \text{for correct MC/boolean (Tier 2)} \nonumber\\
R_{\text{answer}} &= 2.5 && \text{for substring containment (Tier 3)} \nonumber\\
R_{\text{answer}} &\in [1.25,\, 3.5] && \text{for numeric with $\pm 5\%$ or $\pm 10\%$ tolerance (Tier 4)} \nonumber\\
R_{\text{answer}} &= 0.0 && \text{for no match.}
\end{align}

Format compliance rewards match for `<reasoning>`...`</reasoning>` followed by `<answer>`...`</answer>` and a small penalty for missing IDEA-E section marker within `<reasoning>`: [I], [D], [A], and two [E] markers:
\begin{align}
&& R_{\text{format}} &= 0.75 \quad \text{for an exact match of the required structure} \nonumber\\
&& R_{\text{format}} &\pm 0.0625 \quad \text{per required outer tag, i.e., `<reasoning>` and `<answer>`} \nonumber\\
&& R_{\text{IDEA-E}} &= -0.05 \times (\text{number of section tags missing, max 5}) \nonumber\\
\end{align}

### 6.1 TF-IDF Reasoning Similarity
To reward explanation quality beyond correctness, cosine similarity is computed between TF-IDF vectors of teacher and student reasoning traces, yielding values in the range $[0,\, 2.25]$:
\begin{equation}
R_{\text{reasoning}} = 2.25 \cdot \cos\!\bigl(r_{\text{teacher}},\, r_{\text{student}}\bigr),
\end{equation}
where $r_{\text{teacher}}$ represents globalized TF-IDF teacher reasoning traces.

---

````python
# =============================================================================
# REWARD FUNCTIONS
# =============================================================================
match_format = re.compile(
    rf"^[\s]{{0,}}"
    rf"{REASONING_START}.+?{REASONING_END}.*?"
    rf"{SOLUTION_START}(.+?){SOLUTION_END}"
    rf"[\s]{{0,}}$",
    flags=re.MULTILINE | re.DOTALL,
)

def match_format_exactly(prompts, completions, **kwargs):
    """Reward 0.75 for exact format match."""
    return [
        0 if match_format.search(response) is None else 0.75
        for response in completions
    ]

def match_format_approximately(prompts, completions, **kwargs):
    """Partial credit for format tags."""
    scores = []
    for completion in completions:
        score = 0
        score += 0.0625 if completion.count(REASONING_START) == 1 else -0.0625
        score += 0.0625 if completion.count(REASONING_END) == 1 else -0.0625
        score += 0.0625 if completion.count(SOLUTION_START) == 1 else -0.0625
        score += 0.0625 if completion.count(SOLUTION_END) == 1 else -0.0625
        scores.append(score)
    return scores

def check_ideae_sections(prompts, completions, **kwargs):
    """Reward for IDEA-E section markers [I], [D], [E], [A], [E] in reasoning."""
    scores = []
    for completion in completions:
        score = 0.0
        if "[I" not in completion:
            score -= 0.05
        if "[D" not in completion:
            score -= 0.05
        if "[A" not in completion:
            score -= 0.05
        e_count = completion.count("[E")
        if e_count < 1:
            score -= 0.05
        if e_count < 2:
            score -= 0.05
        scores.append(score)
    return scores

def check_answer(prompts, completions, answer, **kwargs):
    """TF-IDF reasoning similarity reward."""
    gt_reasoning = kwargs.get("reasoning", [])
    reasoning_block_pattern = re.compile(r"<reasoning>(.+?)</reasoning>", re.DOTALL)
    
    scores = []
    for completion, gt in zip(completions, gt_reasoning):
        if not gt:
            scores.append(0.0)
            continue
        reasoning_match = reasoning_block_pattern.search(completion)
        if not reasoning_match:
            scores.append(0.0)
            continue
        model_reasoning = reasoning_match.group(1)
        try:
            vecs = reasoning_tfidf.transform([gt, model_reasoning])
            sim = cosine_similarity(vecs[0:1], vecs[1:2])[0][0]
            score = sim * 2.25
            scores.append(score)
        except Exception:
            scores.append(0.0)
    return scores

def extract_mc_letter(text):
    """Extract multiple choice letter (A-E) from answer text."""
    patterns = [
        r'^\s*([A-Ea-e])\s*[:\.\)]',
        r'^\s*([A-Ea-e])\s*$',
        r'answer\s+is\s+([A-Ea-e])',
        r'^\s*([A-Ea-e])\s+',
    ]
    for pattern in patterns:
        match = re.search(pattern, text, re.IGNORECASE)
        if match:
            return match.group(1).lower()
    return None

def normalize_boolean(text):
    """Normalize yes/no, true/false to boolean."""
    text = text.strip().lower()
    if text in ['yes', 'true', 'correct', 'right', 'wrong', 'morally wrong']:
        return True
    if text in ['no', 'false', 'incorrect', 'not wrong', 'not morally wrong']:
        return False
    if 'wrong' in text and 'not wrong' not in text:
        return True
    if 'not wrong' in text:
        return False
    if text.startswith('yes') or text.startswith('yes,'):
        return True
    if text.startswith('no ') or text.startswith('no,'):
        return False
    return None

def extract_number(text):
    """Extract first number from text."""
    cleaned = re.sub(r'[$,]', '', text)
    match = re.search(r'-?\d+\.?\d*', cleaned)
    if match:
        return float(match.group())
    return None

def check_numbers(prompts, completions, answer, **kwargs):
    """Enhanced answer checking with MC, boolean, and number support."""
    answer_pattern = re.compile(rf"{SOLUTION_START}(.+?){SOLUTION_END}", re.DOTALL)
    extracted_responses = [
        m.group(1).strip() if (m := answer_pattern.search(r)) else None
        for r in completions
    ]
    
    scores = []
    for guess, true_answer in zip(extracted_responses, answer):
        if guess is None or true_answer is None:
            scores.append(0)
            continue
        
        guess_clean = guess.strip().lower()
        true_clean = true_answer.strip().lower()
        
        # TIER 1: Exact text match
        if guess_clean == true_clean:
            scores.append(4.0)
            continue
        
        # TIER 2: Multiple choice letter
        guess_letter = extract_mc_letter(guess_clean)
        true_letter = extract_mc_letter(true_clean)
        if guess_letter and true_letter:
            scores.append(3.5 if guess_letter == true_letter else 0)
            continue
        
        # TIER 3: Boolean normalization
        guess_bool = normalize_boolean(guess_clean)
        true_bool = normalize_boolean(true_clean)
        if guess_bool is not None and true_bool is not None:
            scores.append(3.5 if guess_bool == true_bool else 0)
            continue
        
        # TIER 4: Substring containment
        if true_clean in guess_clean:
            scores.append(2.5)
            continue
        
        # TIER 5: Numeric comparison
        guess_num = extract_number(guess_clean)
        true_num = extract_number(true_clean)
        if guess_num is not None and true_num is not None:
            if guess_num == true_num:
                scores.append(3.5)
            elif true_num != 0:
                ratio = guess_num / true_num
                if 0.95 <= ratio <= 1.05:
                    scores.append(2.5)
                elif 0.9 <= ratio <= 1.1:
                    scores.append(1.25)
                else:
                    scores.append(0)
            else:
                scores.append(0)
            continue
        
        scores.append(0)
    
    return scores
````

````python
# =============================================================================
# MODEL LOADING
# =============================================================================

# Download model
model_path = {"gemma2": "google/gemma-2/flax/"}
model_family = "gemma2"
model_version = "gemma2-2b-it"
print(f"{model_path[model_family]}{model_version}")

kaggle_ckpt_path = kagglehub.model_download(
    f"{model_path[model_family]}{model_version}"
)

# Convert checkpoint format
if model_family == "gemma2":
    params = params_lib.load_and_format_params(
        os.path.join(kaggle_ckpt_path, "gemma2-2b-it")
    )
    gemma = gemma_lib.Transformer.from_params(params, version="2-2b-it")
    
    # Clear intermediate checkpoint directory if it exists
    intermediate_state_path = os.path.join(INTERMEDIATE_CKPT_DIR, "state")
    if os.path.exists(intermediate_state_path):
        shutil.rmtree(intermediate_state_path)
        print(f"Cleared existing checkpoint at {intermediate_state_path}")
    
    checkpointer = ocp.StandardCheckpointer()
    _, state = nnx.split(gemma)
    checkpointer.save(intermediate_state_path, state)
    checkpointer.wait_until_finished()
    del params, gemma, state
    gc.collect()


def get_gemma_ref_model(ckpt_path):
    mesh = jax.make_mesh(*MESH)
    model_config = gemma_lib.ModelConfig.gemma2_2b()
    abs_gemma = nnx.eval_shape(
        lambda: gemma_lib.Transformer(model_config, rngs=nnx.Rngs(params=0))
    )
    abs_state = nnx.state(abs_gemma)
    abs_state = jax.tree.map(
        lambda a, s: jax.ShapeDtypeStruct(a.shape, jnp.bfloat16, sharding=s),
        abs_state,
        nnx.get_named_sharding(abs_state, mesh),
    )
    checkpointer = ocp.StandardCheckpointer()
    restored_params = checkpointer.restore(ckpt_path, target=abs_state)
    graph_def, _ = nnx.split(abs_gemma)
    gemma = nnx.merge(graph_def, restored_params)
    return gemma, mesh, model_config


def get_lora_model(base_model, mesh):
    lora_provider = qwix.LoraProvider(
        module_path=(
            ".*q_einsum|.*kv_einsum|.*gate_proj|.*down_proj|.*up_proj|"
            ".*attn_vec_einsum"
        ),
        rank=RANK,
        alpha=ALPHA,
    )
    model_input = base_model.get_model_input()
    lora_model = qwix.apply_lora_to_model(base_model, lora_provider, **model_input)
    with mesh:
        state = nnx.state(lora_model)
        pspecs = nnx.get_partition_spec(state)
        sharded_state = jax.lax.with_sharding_constraint(state, pspecs)
        nnx.update(lora_model, sharded_state)
    return lora_model


# Load reference and policy models
ref_model, mesh, model_config = get_gemma_ref_model(
    ckpt_path=os.path.join(INTERMEDIATE_CKPT_DIR, "state")
)
lora_policy = get_lora_model(ref_model, mesh=mesh)
nnx.display(lora_policy)

# Load tokenizer
tokenizer = tokenizer_lib.Tokenizer(
    tokenizer_path=os.path.join(kaggle_ckpt_path, "tokenizer.model")
)
EOS_TOKENS = [tokenizer.eos_id()]
print(f"Using EOS token IDs: {EOS_TOKENS}")
````

````python
# =============================================================================
# EVALUATION HELPERS
# =============================================================================

def create_sampler(model):
    """Create a sampler for generation."""
    return sampler_lib.Sampler(
        transformer=model,
        tokenizer=tokenizer,
        cache_config=sampler_lib.CacheConfig(
            cache_size=MAX_PROMPT_LENGTH + TOTAL_GENERATION_STEPS + 256,
            num_layers=model_config.num_layers,
            num_kv_heads=model_config.num_kv_heads,
            head_dim=model_config.head_dim,
        ),
    )

def generate(question, sampler, temperature=0.7, top_k=50, top_p=0.95, seed=None):
    """Generate response for a question.
    
    For greedy decoding, pass temperature=None, top_k=1, top_p=None.
    """
    if isinstance(question, str):
        question = [question]
    
    prompts = [PROMPT_TEMPLATE.format(question=q) for q in question]
    
    # Sampler is called directly (not .generate())
    out_data = sampler(
        input_strings=prompts,
        max_generation_steps=TOTAL_GENERATION_STEPS,
        temperature=temperature,
        top_k=top_k,
        top_p=top_p,
        echo=False,
        seed=seed if seed is not None else None,
    )
    
    output = out_data.text
    if len(question) == 1:
        return output[0]
    return output

def evaluate_sample(questions, answers, sampler, num_samples=20, verbose=True, seed=42):
    """Evaluate model on a sample of questions. Uses fixed seed for reproducibility."""
    rng = np.random.default_rng(seed)
    indices = rng.choice(len(questions), min(num_samples, len(questions)), replace=False)
    
    correct = 0
    format_ok = 0
    ideae_ok = 0
    
    for i, idx in enumerate(indices):
        q = questions[idx]
        true_ans = answers[idx]
        
        # Greedy decoding: temperature=None, top_p=None
        response = generate(q, sampler, temperature=None, top_k=1, top_p=None)
        
        # Check format
        fmt_match = match_format.search(response) is not None
        if fmt_match:
            format_ok += 1
        
        # Check IDEA-E sections
        has_ideae = all(tag in response for tag in ["[I", "[D", "[A"]) and response.count("[E") >= 2
        if has_ideae:
            ideae_ok += 1
        
        # Check answer
        answer_match = re.search(rf"{SOLUTION_START}(.+?){SOLUTION_END}", response, re.DOTALL)
        extracted = answer_match.group(1).strip() if answer_match else "[NO ANSWER BLOCK]"
        true_clean = str(true_ans).strip().lower()
        
        is_correct = False
        if answer_match:
            extracted_lower = extracted.lower()
            # Check various match types
            if extracted_lower == true_clean:
                correct += 1
                is_correct = True
            elif extract_mc_letter(extracted_lower) and extract_mc_letter(extracted_lower) == extract_mc_letter(true_clean):
                correct += 1
                is_correct = True
            elif normalize_boolean(extracted_lower) is not None and normalize_boolean(extracted_lower) == normalize_boolean(true_clean):
                correct += 1
                is_correct = True
            elif true_clean in extracted_lower:
                correct += 1
                is_correct = True
        
        # Print generation details
        if verbose:
            status = "✓" if is_correct else "✗"
            fmt_status = "F✓" if fmt_match else "F✗"
            ideae_status = "I✓" if has_ideae else "I✗"
            print(f"\n{'='*60}")
            print(f"[{i+1}/{len(indices)}] {status} {fmt_status} {ideae_status}")
            print(f"Q: {q}")
            print(f"TRUE: {true_ans}")
            print(f"PRED: {extracted}")
            print(f"\n--- FULL RESPONSE ---")
            print(response)
    
    n = len(indices)
    return {
        "accuracy": correct / n * 100,
        "format": format_ok / n * 100,
        "ideae": ideae_ok / n * 100,
        "n_samples": n
    }
````

## 6. Discussion and Conclusion

### 6.1 Structured Reasoning as a Training Target

This work demonstrates that structured reasoning formats like IDEA-E can be reliably instilled through reinforcement learning. All trained variants achieved high format and IDEA-E compliance after sufficient training, and final answer correctness improved when utilizing TF-IDF rewards.

### 6.2 Safety Through Structure

The explicit Ethics step [E] ensures safety considerations are evaluated before the Act step, providing a built-in checkpoint for potentially harmful queries. While formal safety benchmarking was not conducted, qualitative review shows consistent presence of ethics reasoning.

### 6.3 Practical Implications

For deployment, consistent output structure simplifies downstream processing and enables systematic auditing. The separation of `<reasoning>` and `<answer>` blocks allows targeted evaluation of reasoning quality independent of answer correctness. The critical role of TF-IDF reasoning rewards suggests that supervision signals beyond task outcomes are necessary for learning structured reasoning behaviors.

### 6.4 Limitations and Future Work

Our experiments were conducted under compute constraints (Colab single-chip TPU for ablations, limited Kaggle TPU time for final runs). Future work should evaluate on broader, held-out distributions and investigate more sophisticated reasoning quality metrics beyond TF-IDF similarity.

### 6.5 Code and Dataset Availability

Training pipeline, dataset, curriculum sampler, and evaluation harness are available at: https://github.com/kpower7/tunix_hackathon. Datasets are publicly available on Kaggle with appropriate license splits: CC-BY-SA-3.0, CC-BY-SA-4.0, and MIT licenses (total 27,198 samples).

````python
# =============================================================================
# MODEL LOADING
# =============================================================================

# Download model
model_path = {"gemma2": "google/gemma-2/flax/"}
model_family = "gemma2"
model_version = "gemma2-2b-it"
print(f"{model_path[model_family]}{model_version}")

kaggle_ckpt_path = kagglehub.model_download(
    f"{model_path[model_family]}{model_version}"
)

# Convert checkpoint format
if model_family == "gemma2":
    params = params_lib.load_and_format_params(
        os.path.join(kaggle_ckpt_path, "gemma2-2b-it")
    )
    gemma = gemma_lib.Transformer.from_params(params, version="2-2b-it")
    
    # Clear intermediate checkpoint directory if it exists
    intermediate_state_path = os.path.join(INTERMEDIATE_CKPT_DIR, "state")
    if os.path.exists(intermediate_state_path):
        shutil.rmtree(intermediate_state_path)
        print(f"Cleared existing checkpoint at {intermediate_state_path}")
    
    checkpointer = ocp.StandardCheckpointer()
    _, state = nnx.split(gemma)
    checkpointer.save(intermediate_state_path, state)
    checkpointer.wait_until_finished()
    del params, gemma, state
    gc.collect()


def get_gemma_ref_model(ckpt_path):
    mesh = jax.make_mesh(*MESH)
    model_config = gemma_lib.ModelConfig.gemma2_2b()
    abs_gemma = nnx.eval_shape(
        lambda: gemma_lib.Transformer(model_config, rngs=nnx.Rngs(params=0))
    )
    abs_state = nnx.state(abs_gemma)
    abs_state = jax.tree.map(
        lambda a, s: jax.ShapeDtypeStruct(a.shape, jnp.bfloat16, sharding=s),
        abs_state,
        nnx.get_named_sharding(abs_state, mesh),
    )
    checkpointer = ocp.StandardCheckpointer()
    restored_params = checkpointer.restore(ckpt_path, target=abs_state)
    graph_def, _ = nnx.split(abs_gemma)
    gemma = nnx.merge(graph_def, restored_params)
    return gemma, mesh, model_config


def get_lora_model(base_model, mesh):
    lora_provider = qwix.LoraProvider(
        module_path=(
            ".*q_einsum|.*kv_einsum|.*gate_proj|.*down_proj|.*up_proj|"
            ".*attn_vec_einsum"
        ),
        rank=RANK,
        alpha=ALPHA,
    )
    model_input = base_model.get_model_input()
    lora_model = qwix.apply_lora_to_model(base_model, lora_provider, **model_input)
    with mesh:
        state = nnx.state(lora_model)
        pspecs = nnx.get_partition_spec(state)
        sharded_state = jax.lax.with_sharding_constraint(state, pspecs)
        nnx.update(lora_model, sharded_state)
    return lora_model


# Load reference and policy models
ref_model, mesh, model_config = get_gemma_ref_model(
    ckpt_path=os.path.join(INTERMEDIATE_CKPT_DIR, "state")
)
lora_policy = get_lora_model(ref_model, mesh=mesh)
nnx.display(lora_policy)

# Load tokenizer
tokenizer = tokenizer_lib.Tokenizer(
    tokenizer_path=os.path.join(kaggle_ckpt_path, "tokenizer.model")
)
EOS_TOKENS = [tokenizer.eos_id()]
print(f"Using EOS token IDs: {EOS_TOKENS}")
````
